```python
def add_messages(messages, text):
    message = {"role": "user", "content": text}
    messages.append(message)

def add_assistant_messages(messages, text):
    message = {"role": "assistant", "content": text}
    messages.append(message)

def chat(messages):
    message = client.messages.create(
    model=model,
    max_tokens=50,
    messages=messages)
    return message.content[0].text
```

```python
messages = []

add_messages(messages, "What is Quantum Computing? Answer it in one sentence")

answer = chat(messages)

print(answer)

add_assistant_messages(messages, answer)

add_messages(messages, "Write Another Sentence")

answer1 = chat(messages)

print(answer1)
```

```python
messages=[]
while True:
    print("Add anything in the input box to chat. Type exit to close the chat")
    user_input = input()
    if (user_input == "exit"):
        break
    print (user_input)
    add_messages(messages, user_input)
    answer = chat(messages)
    add_assistant_messages(messages, answer)
    print("------")
    print(answer)
    print("------")
```

```python
def chat_system(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 50,
        "messages": messages
    }
    if system:
        params["system"] = system
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages=[]

add_messages(messages, "Solve 5x+50=20 for x?")

# Without system prompt
answer = chat_system(messages)

print(answer)

# With system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
answer1 = chat_system(messages, system=system)
print(answer1)
```

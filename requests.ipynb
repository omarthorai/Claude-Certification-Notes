````python
# Create an API Request
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5-20251001"
````

````python
def add_messages(messages, text):
    message = {"role": "user", "content": text}
    messages.append(message)

def add_assistant_messages(messages, text):
    message = {"role": "assistant", "content": text}
    messages.append(message)

def chat(messages):
    message = client.messages.create(
    model=model,
    max_tokens=50,
    messages=messages)
    return message.content[0].text
````

````python
messages = []

add_messages(messages, "What is Quantum Computing? Answer it in one sentence")

answer = chat(messages)

print(answer)

add_assistant_messages(messages, answer)

add_messages(messages, "Write Another Sentence")

answer1 = chat(messages)

print(answer1)
````

````python
messages=[]
while True:
    print("Add anything in the input box to chat. Type exit to close the chat")
    user_input = input()
    if (user_input == "exit"):
        break
    print (user_input)
    add_messages(messages, user_input)
    answer = chat(messages)
    add_assistant_messages(messages, answer)
    print("------")
    print(answer)
    print("------")
````

````python
def chat_system(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 50,
        "messages": messages
    }
    if system:
        params["system"] = system
    message = client.messages.create(**params)
    return message.content[0].text
````

````python
messages=[]

add_messages(messages, "Solve 5x+50=20 for x?")

# Without system prompt
answer = chat_system(messages)

print(answer)

# With system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
answer1 = chat_system(messages, system=system)
print(answer1)
````

````python
messages = []

add_messages(messages, "Write a 1 line description for a fake database")

stream = client.messages.create(
    model = model,
    max_tokens = 50,
    messages = messages,
    stream=True
)

for event in stream:
    print(event)
````

````python
messages = []

add_messages(messages, "Write a 1 line description for a fake database")

with client.messages.stream(
    model=model,
    max_tokens=50,
    messages=messages
) as stream:
    for text in stream.text_stream:
        print(text, end="")

stream.get_final_message()
````

### Streaming Code Explaination

#### `client.messages.create()`
This is the basic call. You send your request, and then you wait until Claude finishes writing the entire response. Only then do you get it back, all at once.

* Good for: simple scripts, short responses, or when you don't need to show anything until it's done.
* You can still get a stream of events from `create()` by passing `stream=True`, but it just hands you the raw event chunks — you'd have to assemble the final message yourself.

#### `client.messages.stream()`
This is a helper built specifically for streaming. Instead of waiting for the whole response, you get the text as it's generated, piece by piece — like watching someone type in real time.

* It's a context manager (used with `with ... as stream:`), and it gives you convenient extras:
   * `stream.text_stream` — iterate over just the text as it arrives
   * Event handlers for different event types
   * Automatic accumulation into a final message object once streaming finishes, if you need it

#### `with ... as`: Context Manager
This is a Python pattern called a context manager. It handles setup and cleanup automatically:

* **Setup**: When the `with` block starts, it opens the streaming connection.
* **Cleanup**: When the block ends (whether it finishes normally or an error happens), it automatically closes the connection for you.

Without `with`, you'd have to manually remember to close the stream/connection yourself every time.

Streaming keeps a connection open the whole time data is coming in. If you don't properly close it:

* You could leave network connections hanging open
* You might hit connection limits on repeated use
* Resources on your machine or Claude's API usage could be wasted

The `with` block guarantees cleanup happens, even if your code crashes inside the loop.

#### What `as stream` gives you
This names the opened connection as `stream`, so you can interact with it inside the block:

* `stream.text_stream` → iterate over just the text pieces as they arrive
* `stream.get_final_message()` → (after the loop) get the complete assembled message, if you need the whole thing too
* Other event-handling attributes, depending on what you need
